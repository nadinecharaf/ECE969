# ScienceQA Ablation Study — Analysis
**ECE 9660 · Western University**

Models: `llama3.2:3b`, `phi4-mini`, `llama3.1:8b`  
Conditions: C1 (Lecture) · C2 (+Hint) · C3 (+Grade) · C4 (Full)  
Sample: 300 examples from ScienceQA test split, stratified by grade, text-only

```python
import json
import glob
import sys
import os

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

# Make sure project root is on path
sys.path.insert(0, os.path.abspath('..'))

plt.rcParams.update({
    'figure.dpi': 130,
    'font.family': 'DejaVu Sans',
    'axes.spines.top': False,
    'axes.spines.right': False,
})

PURPLE   = '#4F2683'   # Western purple
ORANGE   = '#E67E22'
GREEN    = '#27AE60'
BLUE     = '#2980B9'
MODEL_COLORS = {
    'llama3.2-3b':  BLUE,
    'phi4-mini':    ORANGE,
    'llama3.1-8b':  PURPLE,
}
COND_LABELS = {
    'C1': 'C1\nBaseline',
    'C2': 'C2\n+Hint',
    'C3': 'C3\n+Grade',
    'C4': 'C4\nFull',
}

TARGET_MODELS = {'llama3.2-3b', 'phi4-mini', 'llama3.1-8b'}
RESULTS_DIR   = '../results'
print('Setup complete.')
```

## 1 · Load Results

```python
rows = []
seen = {}   # keep only the latest file per (model, condition)

for f in sorted(glob.glob(f'{RESULTS_DIR}/*.json')):
    with open(f, encoding='utf-8') as fp:
        d = json.load(fp)
    model = d.get('model', '')
    cond  = d.get('condition', '')
    n     = d.get('n', len(d.get('results', [])))
    if model not in TARGET_MODELS or n < 100:
        continue
    seen[(model, cond)] = d   # overwrite with latest

for (model, cond), d in seen.items():
    for r in d['results']:
        rows.append(r)

df = pd.DataFrame(rows)
print(f'Total rows loaded : {len(df)}')
print(f'Models            : {sorted(df.model.unique())}')
print(f'Conditions        : {sorted(df.condition.unique())}')
print(f'Grades            : {sorted(df.grade.dropna().unique())}')
df.head(3)
```

## 2 · Main Accuracy Table (Model × Condition)

```python
MODEL_ORDER = ['llama3.2-3b', 'phi4-mini', 'llama3.1-8b']
COND_ORDER  = ['C1', 'C2', 'C3', 'C4']

acc_table = (
    df.groupby(['model', 'condition'])['is_correct']
    .mean()
    .unstack('condition')
    .reindex(index=MODEL_ORDER, columns=COND_ORDER)
    .mul(100)
    .round(1)
)
acc_table.columns.name = None
acc_table.index.name   = 'Model'

# Delta columns — change from C1 baseline
acc_table['ΔC2 vs C1'] = (acc_table['C2'] - acc_table['C1']).round(1)
acc_table['ΔC3 vs C1'] = (acc_table['C3'] - acc_table['C1']).round(1)
acc_table['ΔC4 vs C1'] = (acc_table['C4'] - acc_table['C1']).round(1)

print('Accuracy (%) and delta vs C1 baseline:')
acc_table
```

## 3 · Figure 1 — Accuracy by Condition per Model

```python
fig, ax = plt.subplots(figsize=(8, 4.5))

x      = np.arange(len(COND_ORDER))
width  = 0.25
offset = [-1, 0, 1]

for i, model in enumerate(MODEL_ORDER):
    vals = [acc_table.loc[model, c] for c in COND_ORDER]
    bars = ax.bar(x + offset[i] * width, vals,
                  width=width, color=MODEL_COLORS[model],
                  label=model, alpha=0.88, zorder=3)
    for bar, v in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.4,
                f'{v:.1f}', ha='center', va='bottom', fontsize=7.5)

ax.set_xticks(x)
ax.set_xticklabels(['C1\nBaseline', 'C2\n+Hint', 'C3\n+Grade', 'C4\nFull Context'])
ax.set_ylabel('Accuracy (%)')
ax.set_title('Answer Accuracy by Condition and Model', fontweight='bold', pad=10)
ax.set_ylim(60, 95)
ax.yaxis.set_minor_locator(mticker.MultipleLocator(2))
ax.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
ax.legend(title='Model', bbox_to_anchor=(1.01, 1), loc='upper left', fontsize=9)

plt.tight_layout()
plt.savefig('../results/fig1_accuracy_by_condition.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved → results/fig1_accuracy_by_condition.png')
```

## 4 · Figure 2 — Accuracy Heatmap (Model × Condition)

```python
fig, ax = plt.subplots(figsize=(6, 3))

heat_data = acc_table[COND_ORDER]
sns.heatmap(
    heat_data,
    annot=True, fmt='.1f',
    cmap='YlOrRd',
    linewidths=0.5,
    ax=ax,
    vmin=60, vmax=90,
    cbar_kws={'label': 'Accuracy (%)'}
)
ax.set_title('Accuracy Heatmap (Model × Condition)', fontweight='bold')
ax.set_xlabel('Condition')
ax.set_ylabel('Model')

plt.tight_layout()
plt.savefig('../results/fig2_accuracy_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved → results/fig2_accuracy_heatmap.png')
```

## 5 · Figure 3 — Accuracy by Grade (per Model, C1 vs C4)

```python
grade_order = [f'grade{i}' for i in range(1, 13)]
grade_labels = [f'G{i}' for i in range(1, 13)]

fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)

for ax, model in zip(axes, MODEL_ORDER):
    sub = df[df.model == model]
    c1  = sub[sub.condition == 'C1'].groupby('grade')['is_correct'].mean().reindex(grade_order) * 100
    c4  = sub[sub.condition == 'C4'].groupby('grade')['is_correct'].mean().reindex(grade_order) * 100

    x = np.arange(len(grade_order))
    ax.bar(x - 0.2, c1, width=0.38, label='C1 Baseline', color=BLUE,   alpha=0.8, zorder=3)
    ax.bar(x + 0.2, c4, width=0.38, label='C4 Full',     color=PURPLE, alpha=0.8, zorder=3)

    ax.set_title(model, fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels(grade_labels, fontsize=8)
    ax.set_ylim(0, 105)
    ax.set_xlabel('Grade')
    ax.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
    ax.legend(fontsize=8)

axes[0].set_ylabel('Accuracy (%)')
fig.suptitle('Per-Grade Accuracy: C1 (Baseline) vs C4 (Full Context)', fontweight='bold', y=1.02)

plt.tight_layout()
plt.savefig('../results/fig3_accuracy_by_grade.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved → results/fig3_accuracy_by_grade.png')
```

## 6 · Figure 4 — Accuracy by Subject

```python
subjects = sorted(df.subject.dropna().unique())

fig, ax = plt.subplots(figsize=(8, 4))

x     = np.arange(len(subjects))
width = 0.25

for i, model in enumerate(MODEL_ORDER):
    sub  = df[(df.model == model) & (df.condition == 'C4')]
    vals = [sub[sub.subject == s]['is_correct'].mean() * 100 for s in subjects]
    bars = ax.bar(x + (i - 1) * width, vals,
                  width=width, color=MODEL_COLORS[model],
                  label=model, alpha=0.88, zorder=3)
    for bar, v in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.4,
                f'{v:.0f}', ha='center', va='bottom', fontsize=8)

ax.set_xticks(x)
ax.set_xticklabels(subjects)
ax.set_ylabel('Accuracy (%) — C4 Full Context')
ax.set_title('Accuracy by Subject (C4 Full Context)', fontweight='bold')
ax.set_ylim(50, 100)
ax.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
ax.legend(title='Model', bbox_to_anchor=(1.01, 1), loc='upper left', fontsize=9)

plt.tight_layout()
plt.savefig('../results/fig4_accuracy_by_subject.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved → results/fig4_accuracy_by_subject.png')
```

## 7 · Choice Consistency (Instruction Following)

```python
consistency = (
    df.assign(parseable=df['predicted_answer'].notna())
    .groupby(['model', 'condition'])['parseable']
    .mean()
    .unstack('condition')
    .reindex(index=MODEL_ORDER, columns=COND_ORDER)
    .mul(100)
    .round(1)
)
print('Choice Consistency (%) — fraction of responses with a parseable answer:')
consistency
```

## 8 · RQ Analysis — Isolating Each Effect

```python
print('=== RQ2: Hint Effect (C2 - C1) ===')
print('Positive = hints help accuracy')
print(acc_table[['C1','C2','ΔC2 vs C1']].to_string())
print()
print('=== RQ3: Grade Context Effect (C3 - C1) ===')
print('Positive = grade metadata helps accuracy')
print(acc_table[['C1','C3','ΔC3 vs C1']].to_string())
print()
print('=== Full Context Effect (C4 - C1) ===')
print('Positive = all context together helps')
print(acc_table[['C1','C4','ΔC4 vs C1']].to_string())
```

```python
# Hint effect: split by whether the example actually has a hint
print('=== RQ2 refined: Hint effect split by has_hint ===')
for model in MODEL_ORDER:
    sub = df[df.model == model]
    c1_with    = sub[(sub.condition=='C1') & (sub.has_hint==True) ]['is_correct'].mean()
    c1_without = sub[(sub.condition=='C1') & (sub.has_hint==False)]['is_correct'].mean()
    c2_with    = sub[(sub.condition=='C2') & (sub.has_hint==True) ]['is_correct'].mean()
    c2_without = sub[(sub.condition=='C2') & (sub.has_hint==False)]['is_correct'].mean()
    print(f'\n{model}:')
    print(f'  has_hint=True  : C1={c1_with:.3f}  C2={c2_with:.3f}  delta={c2_with-c1_with:+.3f}')
    print(f'  has_hint=False : C1={c1_without:.3f}  C2={c2_without:.3f}  delta={c2_without-c1_without:+.3f}')
```

## 9 · Figure 5 — Condition Effect (Delta from C1)

```python
delta_cols = ['ΔC2 vs C1', 'ΔC3 vs C1', 'ΔC4 vs C1']
delta_data  = acc_table[delta_cols].copy()
delta_data.columns = ['+Hint (C2)', '+Grade (C3)', 'Full (C4)']

fig, ax = plt.subplots(figsize=(7, 4))

x     = np.arange(3)
width = 0.25

for i, model in enumerate(MODEL_ORDER):
    vals = delta_data.loc[model].values
    bars = ax.bar(x + (i - 1) * width, vals,
                  width=width, color=MODEL_COLORS[model],
                  label=model, alpha=0.88, zorder=3)
    for bar, v in zip(bars, vals):
        ypos = bar.get_height() + 0.1 if v >= 0 else bar.get_height() - 0.6
        ax.text(bar.get_x() + bar.get_width()/2, ypos,
                f'{v:+.1f}', ha='center', va='bottom', fontsize=8)

ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
ax.set_xticks(x)
ax.set_xticklabels(['+Hint\n(C2 vs C1)', '+Grade\n(C3 vs C1)', 'Full Context\n(C4 vs C1)'])
ax.set_ylabel('Accuracy Change (pp)')
ax.set_title('Condition Effect Relative to C1 Baseline', fontweight='bold')
ax.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
ax.legend(title='Model', bbox_to_anchor=(1.01, 1), loc='upper left', fontsize=9)

plt.tight_layout()
plt.savefig('../results/fig5_condition_delta.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved → results/fig5_condition_delta.png')
```

## 10 · Qualitative Analysis

```python
def show_example(row, label=''):
    choices = row['choices']
    labels  = ['A','B','C','D','E'][:len(choices)]
    gold_letter  = labels[int(row['gold_answer'])]
    pred_idx     = row['predicted_answer']
    pred_letter  = labels[int(pred_idx)] if pred_idx is not None else '?'
    correct_str  = 'CORRECT' if row['is_correct'] else 'WRONG'

    print(f'--- {label} [{correct_str}] ---')
    print(f'Grade   : {row["grade"]}  |  Subject: {row["subject"]}')
    print(f'Question: {row["question"]}')
    for l, c in zip(labels, choices):
        marker = '<-- correct' if l == gold_letter else ''
        print(f'  {l}. {c}  {marker}')
    print(f'Predicted: {pred_letter}  |  Gold: {gold_letter}')
    if row.get('lecture'):
        print(f'Lecture (first 200 chars): {row["lecture"][:200]}...')
    print(f'Response (first 400 chars):')
    print(row['response'][:400])
    print()
```

```python
# Best model, correct answers — examples of good grounding
best_model   = 'llama3.1-8b'
good_examples = df[
    (df.model == best_model) &
    (df.condition == 'C4') &
    (df.is_correct == True)
].sample(3, random_state=42)

print('=== CORRECT ANSWERS — llama3.1-8b, C4 Full Context ===')
for _, row in good_examples.iterrows():
    show_example(row, f'C4/{best_model}')
```

```python
# Failure cases — wrong answers
fail_examples = df[
    (df.model == best_model) &
    (df.condition == 'C1') &
    (df.is_correct == False)
].sample(3, random_state=42)

print('=== FAILURE CASES — llama3.1-8b, C1 Baseline ===')
for _, row in fail_examples.iterrows():
    show_example(row, f'C1/{best_model}')
```

```python
# Grade conditioning: same question, C1 vs C4 response for a low grade
low_grade_q = df[
    (df.model == best_model) &
    (df.grade.isin(['grade1','grade2','grade3'])) &
    (df.is_correct == True)
]

# Find a question answered correctly in both C1 and C4
c1_q  = set(low_grade_q[low_grade_q.condition == 'C1']['question'])
c4_q  = set(low_grade_q[low_grade_q.condition == 'C4']['question'])
common = list(c1_q & c4_q)

if common:
    q = common[0]
    r_c1 = df[(df.model==best_model)&(df.condition=='C1')&(df.question==q)].iloc[0]
    r_c4 = df[(df.model==best_model)&(df.condition=='C4')&(df.question==q)].iloc[0]
    print('=== GRADE CONDITIONING COMPARISON (same question, C1 vs C4) ===')
    print(f'Grade: {r_c1["grade"]}  |  Subject: {r_c1["subject"]}')
    print(f'Question: {q}')
    print()
    print('--- C1 (no grade info) response ---')
    print(r_c1['response'][:500])
    print()
    print('--- C4 (with grade info) response ---')
    print(r_c4['response'][:500])
else:
    print('No matching examples found — try different grade filters.')
```

## 11 · Summary of Findings

```python
print('=' * 60)
print('SUMMARY OF FINDINGS')
print('=' * 60)

print('\n[Model Size Effect]')
c4_acc = acc_table['C4']
for m in MODEL_ORDER:
    print(f'  {m:<22}: {c4_acc[m]:.1f}%  (C4 Full Context)')

print('\n[RQ2 - Hint Effect: C2 vs C1]')
for m in MODEL_ORDER:
    d = acc_table.loc[m, 'ΔC2 vs C1']
    print(f'  {m:<22}: {d:+.1f} pp')

print('\n[RQ3 - Grade Context Effect: C3 vs C1]')
for m in MODEL_ORDER:
    d = acc_table.loc[m, 'ΔC3 vs C1']
    print(f'  {m:<22}: {d:+.1f} pp')

print('\n[Full Context Effect: C4 vs C1]')
for m in MODEL_ORDER:
    d = acc_table.loc[m, 'ΔC4 vs C1']
    print(f'  {m:<22}: {d:+.1f} pp')

print()
print('Choice Consistency (all conditions >= 99%?):')
print(consistency.to_string())
print('=' * 60)
```
