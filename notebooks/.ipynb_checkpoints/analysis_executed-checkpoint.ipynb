# ScienceQA Ablation Study — Analysis
**ECE 9660 · Western University**

Models: `llama3.2:3b` · `phi4-mini` · `llama3.1:8b` · `mixtral:8x7b` · `claude-haiku`  
Conditions: C0 (No Context) · C1 (+Lecture) · C2 (+Hint) · C3 (+Grade) · C4 (Full)  
Sample: 375 examples · grades 3-7 · 75/grade · text-only · all have lecture+hint+solution

```python
import json, glob, sys, os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sys.path.insert(0, os.path.abspath('..'))
plt.rcParams.update({'figure.dpi':130,'font.family':'DejaVu Sans',
                     'axes.spines.top':False,'axes.spines.right':False})
PURPLE='#4F2683'; BLUE='#2980B9'; ORANGE='#E67E22'; GREEN='#27AE60'; RED='#C0392B'
MODEL_ORDER  = ['llama3.2-3b','phi4-mini','llama3.1-8b','mixtral-8x7b','claude-haiku']
MODEL_LABELS = ['Llama 3.2 3B','Phi-4 Mini','Llama 3.1 8B','Mixtral 8x7B','Claude Haiku']
MODEL_COLORS = dict(zip(MODEL_ORDER,[BLUE,ORANGE,PURPLE,GREEN,RED]))
COND_ORDER   = ['C0','C1','C2','C3','C4']
COND_LABELS  = {'C0':'C0\nNo Ctx','C1':'C1\n+Lecture','C2':'C2\n+Hint',
                'C3':'C3\n+Grade','C4':'C4\nFull'}
GRADE_ORDER  = ['grade3','grade4','grade5','grade6','grade7']
GRADE_LABELS = ['Grade 3','Grade 4','Grade 5','Grade 6','Grade 7']
RESULTS_DIR  = '../results'
print('Setup complete.')
```

## 1 · Load Results

```python
rows = []
seen = {}
for f in sorted(glob.glob(f'{RESULTS_DIR}/*.json')):
    with open(f, encoding='utf-8') as fp:
        d = json.load(fp)
    model = d.get('model', '')
    cond  = d.get('condition', '')
    if not model or not cond:
        continue
    seen[(model, cond)] = d
for (model, cond), d in seen.items():
    for r in d['results']:
        rows.append(r)
df = pd.DataFrame(rows)
df['model'] = df['model'].str.replace(
    'claude-haiku-4-5-20251001', 'claude-haiku', regex=False)
df = df[df['model'].isin(MODEL_ORDER) & df['grade'].isin(GRADE_ORDER)]
print(f'Rows: {len(df)}')
print(f'Models: {sorted(df.model.unique())}')
print(f'Conditions: {sorted(df.condition.unique())}')
```

## 2 · Accuracy Table (Model x Condition)

```python
acc = (
    df.groupby(['model','condition'])['is_correct']
    .mean().unstack('condition')
    .reindex(index=MODEL_ORDER, columns=COND_ORDER)
    .mul(100).round(1)
)
acc.index = MODEL_LABELS
acc.columns.name = None
for c in ['C1','C2','C3','C4']:
    acc[f'D{c}'] = (acc[c] - acc['C0']).round(1)
print('Accuracy (%) and delta vs C0:')
display(acc)
```

## 3 · Figure 1 — Accuracy Heatmap (Model x Condition)

```python
fig, ax = plt.subplots(figsize=(7,4))
sns.heatmap(acc[COND_ORDER], annot=True, fmt='.1f', cmap='YlOrRd',
            linewidths=0.5, ax=ax, vmin=60, vmax=100,
            cbar_kws={'label':'Accuracy (%)'})
ax.set_title('Answer Accuracy (%) — Model x Condition', fontweight='bold', pad=10)
ax.tick_params(axis='y', rotation=0)
plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/fig1_accuracy_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig1_accuracy_heatmap.png')
```

## 4 · Figure 2 — Accuracy by Condition (Line Plot)

```python
fig, ax = plt.subplots(figsize=(9,5))
x = np.arange(len(COND_ORDER))
for model, label in zip(MODEL_ORDER, MODEL_LABELS):
    vals = [acc.loc[label, c] for c in COND_ORDER]
    ax.plot(x, vals, marker='o', linewidth=2, markersize=7,
            color=MODEL_COLORS[model], label=label)
    for xi, v in zip(x, vals):
        ax.text(xi, v+0.8, f'{v:.1f}', ha='center', va='bottom',
                fontsize=7.5, color=MODEL_COLORS[model])
ax.set_xticks(x)
ax.set_xticklabels([COND_LABELS[c] for c in COND_ORDER])
ax.set_ylabel('Accuracy (%)')
ax.set_title('Accuracy by Condition per Model', fontweight='bold', pad=10)
ax.set_ylim(55, 108)
ax.grid(axis='y', linestyle='--', alpha=0.4)
ax.legend(bbox_to_anchor=(1.01,1), loc='upper left', fontsize=9)
plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/fig2_accuracy_by_condition.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig2_accuracy_by_condition.png')
```

## 5 · Figure 3 — Condition Effect (Delta vs C0 Baseline)

```python
x = np.arange(4)
w = 0.15
offsets = np.linspace(-2, 2, len(MODEL_ORDER)) * w
fig, ax = plt.subplots(figsize=(9,4.5))
for i, (model, label) in enumerate(zip(MODEL_ORDER, MODEL_LABELS)):
    vals = [acc.loc[label, f'D{c}'] for c in ['C1','C2','C3','C4']]
    bars = ax.bar(x+offsets[i], vals, width=w,
                  color=MODEL_COLORS[model], label=label, alpha=0.88, zorder=3)
    for bar, v in zip(bars, vals):
        ax.text(bar.get_x()+bar.get_width()/2,
                bar.get_height()+(0.3 if v>=0 else -1.2),
                f'{v:+.1f}', ha='center', va='bottom', fontsize=7)
ax.axhline(0, color='black', linewidth=0.8)
ax.set_xticks(x)
ax.set_xticklabels(['+Lecture (C1)','+Hint (C2)','+Grade (C3)','Full (C4)'])
ax.set_ylabel('Accuracy Change (pp vs C0)')
ax.set_title('Condition Effect Relative to C0 Baseline', fontweight='bold', pad=10)
ax.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
ax.legend(bbox_to_anchor=(1.01,1), loc='upper left', fontsize=8)
plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/fig3_condition_delta.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig3_condition_delta.png')
```

## 6 · Figure 4 — Per-Grade Accuracy (C0 vs C2 vs C4)

```python
compare = ['C0','C2','C4']
col = {'C0':BLUE, 'C2':ORANGE, 'C4':PURPLE}
fig, axes = plt.subplots(1, 5, figsize=(18,4), sharey=True)
for ax, model, label in zip(axes, MODEL_ORDER, MODEL_LABELS):
    sub = df[df.model == model]
    x = np.arange(5); w = 0.25; off = [-1, 0, 1]
    for o, cond in zip(off, compare):
        s2 = sub[sub.condition == cond]
        vals = [s2[s2.grade==g]['is_correct'].mean()*100 for g in GRADE_ORDER]
        ax.bar(x+o*w, vals, width=w, color=col[cond], label=cond, alpha=0.85, zorder=3)
    ax.set_title(label, fontweight='bold', fontsize=9)
    ax.set_xticks(x)
    ax.set_xticklabels(['G3','G4','G5','G6','G7'], fontsize=8)
    ax.set_ylim(40, 112)
    ax.grid(axis='y', linestyle='--', alpha=0.4, zorder=0)
    if ax is axes[0]:
        ax.set_ylabel('Accuracy (%)')
        ax.legend(fontsize=8)
fig.suptitle('Per-Grade Accuracy: C0 vs C2 (+Hint) vs C4 (Full)',
             fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/fig4_accuracy_by_grade.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig4_accuracy_by_grade.png')
```

## 7 · Figure 5 — Grade x Condition Heatmap per Model

```python
fig, axes = plt.subplots(1, 5, figsize=(20,3.5))
for ax, model, label in zip(axes, MODEL_ORDER, MODEL_LABELS):
    sub = df[df.model == model]
    heat = (
        sub.groupby(['grade','condition'])['is_correct']
        .mean().unstack('condition')
        .reindex(index=GRADE_ORDER, columns=COND_ORDER)
        .mul(100).round(1)
    )
    heat.index = GRADE_LABELS
    sns.heatmap(heat, annot=True, fmt='.0f', cmap='YlOrRd',
                vmin=50, vmax=100, linewidths=0.4,
                cbar=(ax is axes[-1]), ax=ax, annot_kws={'size':8})
    ax.set_title(label, fontweight='bold', fontsize=9)
    ax.set_xlabel('Condition')
    ax.set_ylabel('')
    ax.tick_params(axis='y', rotation=0)
fig.suptitle('Accuracy (%) by Grade x Condition', fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(f'{RESULTS_DIR}/fig5_grade_condition_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig5_grade_condition_heatmap.png')
```

## 8 · Choice Consistency

```python
cons = (
    df.assign(p=df['predicted_answer'].notna())
    .groupby(['model','condition'])['p'].mean()
    .unstack('condition')
    .reindex(index=MODEL_ORDER, columns=COND_ORDER)
    .mul(100).round(1)
)
cons.index = MODEL_LABELS
print('Choice Consistency (%) -- fraction of responses with parseable answer letter:')
display(cons)
```

## 9 · RQ Analysis

```python
for rq, cond, title in [
    ('RQ0', 'C0', 'Parametric baseline (C0 -- no context)'),
    ('RQ1', 'C1', 'Lecture grounding effect (C1 - C0)'),
    ('RQ2', 'C2', 'Hint effect (C2 - C0)'),
    ('RQ3', 'C3', 'Grade context effect (C3 - C0)'),
]:
    print(f'=== {rq}: {title} ===')
    for label in MODEL_LABELS:
        if cond == 'C0':
            print(f'  {label:<20}: {acc.loc[label, cond]:.1f}%')
        else:
            d = acc.loc[label, f'D{cond}']
            verdict = 'HELPS' if d > 0 else 'HURTS'
            print(f'  {label:<20}: {d:+.1f} pp  ({verdict})')
    print()
```

## 10 · Qualitative Examples

```python
def show(row, lbl=''):
    labs = list('ABCDE')[:len(row['choices'])]
    gold = labs[int(row['gold_answer'])]
    pred = labs[int(row['predicted_answer'])] if (row['predicted_answer'] is not None and row['predicted_answer'] == row['predicted_answer']) else '?'
    tag = 'CORRECT' if row['is_correct'] else 'WRONG'
    print(f'--- {lbl} [{tag}] | {row["grade"]} ---')
    print(f'Q: {row["question"]}')
    for l, c in zip(labs, row['choices']):
        print(f'  {l}. {c}{" <- ans" if l == gold else ""}')
    print(f'Pred: {pred}')
    print(str(row['response'])[:300])
    print()

best = 'claude-haiku'
print('=== CORRECT -- claude-haiku C4 ===')
for _, r in df[(df.model==best)&(df.condition=='C4')&df.is_correct].sample(2, random_state=1).iterrows():
    show(r, 'C4')
print('=== FAILURES -- claude-haiku C0 ===')
for _, r in df[(df.model==best)&(df.condition=='C0')&~df.is_correct].sample(2, random_state=1).iterrows():
    show(r, 'C0')
```

```python
# Grade conditioning: same Grade-3 question, C0 vs C4
sub = df[df.model == 'claude-haiku']
q0 = set(sub[(sub.condition=='C0')&(sub.grade=='grade3')]['question'])
q4 = set(sub[(sub.condition=='C4')&(sub.grade=='grade3')]['question'])
common = list(q0 & q4)
if common:
    q = common[0]
    r0 = sub[(sub.condition=='C0')&(sub.question==q)].iloc[0]
    r4 = sub[(sub.condition=='C4')&(sub.question==q)].iloc[0]
    print(f'Q: {q}\n')
    print('--- C0 (no context) ---')
    print(str(r0['response'])[:400])
    print('\n--- C4 (full context + grade 3) ---')
    print(str(r4['response'])[:400])
```

## 11 · Summary of Key Findings

```python
print('='*65)
print('KEY FINDINGS')
print('='*65)
print('\n[C4 Full Context accuracy by model]')
for label in MODEL_LABELS:
    print(f'  {label:<20}: {acc.loc[label, "C4"]:.1f}%')
print('\n[Most impactful condition per model (delta vs C0)]')
for label in MODEL_LABELS:
    ds = {c: acc.loc[label, f'D{c}'] for c in ['C1','C2','C3','C4']}
    bc = max(ds, key=ds.get)
    print(f'  {label:<20}: {bc} ({ds[bc]:+.1f} pp)')
print('\n[Lecture effect: does lecture help or hurt accuracy?]')
for label in MODEL_LABELS:
    d = acc.loc[label, 'DC1']
    print(f'  {label:<20}: {d:+.1f} pp  ({"HELPS" if d > 0 else "HURTS"})')
print('\n[Grade 6 drop vs Grade 5 -- C4 full context]')
for model, label in zip(MODEL_ORDER, MODEL_LABELS):
    s = df[(df.model==model) & (df.condition=='C4')]
    g5 = s[s.grade=='grade5']['is_correct'].mean()*100
    g6 = s[s.grade=='grade6']['is_correct'].mean()*100
    print(f'  {label:<20}: G5={g5:.1f}% G6={g6:.1f}% drop={g6-g5:+.1f} pp')
print('='*65)
```
